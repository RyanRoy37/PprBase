```python
import json
from docxtpl import DocxTemplate

# Load your template
template_path = "generated_question_template.docx"  # <- your uploaded template filename
output_path = "test_output.docx"

# Load test JSON
with open("test_data.json", "r") as f:
    context = json.load(f)

# Load template
doc = DocxTemplate(template_path)

# Render with the context
doc.render(context)

# Save generated output
doc.save(output_path)

print("Generated:", output_path)
```
